```python
from src.dataset_csv import ImageCSVDataset
import numpy as np
import matplotlib.pyplot as plt
```

```python
TRAIN_IMAGE_PATH = "/cluster/tufts/hugheslab/datasets/IDRID/Images/Training/"
TRAIN_LABEL_PATH = "/cluster/tufts/hugheslab/datasets/IDRID/Labels/training_split.csv"
train_idrid_dataset = ImageCSVDataset(csv_file=TRAIN_LABEL_PATH, root_dir=TRAIN_IMAGE_PATH, transform=None, transforms=None)
```

```python
# Plot some sample images and labels
for i, sample in enumerate(train_idrid_dataset):
        print(i, sample[0].shape, sample[1].shape)

        ax = plt.subplot(1, 4, i + 1)
        plt.tight_layout()
        ax.set_title(f"Label: {sample[1]}")
        ax.axis('off')
        plt.imshow(sample[0])

        if i == 3:
                plt.show()
                break
```
