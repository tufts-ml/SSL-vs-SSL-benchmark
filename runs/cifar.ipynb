```python
def unpickle(file):
    import pickle
    with open(file, 'rb') as fo:
        dict = pickle.load(fo, encoding='bytes')
    return dict
```

```python
file = "/cluster/tufts/hugheslab/datasets/CIFAR100/cifar-100-python/train"
dict = unpickle(file)
print(dict.keys())
print(dict[b'data'].shape)
# Get number of classes in each type of label
print(len(set(dict[b'fine_labels'])))
print(len(set(dict[b'coarse_labels'])))

print(dict[b'data'].shape)
```
