```python
from src.dataset_csv import ImageCSVDataset
import numpy as np
import matplotlib.pyplot as plt
```

```python
IMAGE_PATH = "/cluster/tufts/hugheslab/datasets/chexpert_sample_data/images"
LABEL_PATH = "/cluster/tufts/hugheslab/datasets/chexpert_sample_data/sampled_train.csv"
chexpert_dataset = ImageCSVDataset(csv_file=LABEL_PATH, root_dir=IMAGE_PATH, transform=None, transforms=None)
```
