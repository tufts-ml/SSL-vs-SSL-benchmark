```python
from src.dataset_csv import ImageCSVDataset
import numpy as np
import matplotlib.pyplot as plt
```

```python
IMAGE_PATH = "/cluster/tufts/hugheslab/datasets/IDRID/Images/"
LABEL_PATH = "/cluster/tufts/hugheslab/datasets/chexpert_sample_data"
chexpert_dataset = ImageCSVDataset(csv_file=LABEL_PATH, root_dir=IMAGE_PATH, transform=None, transforms=None)
```

```python
# Plot some sample images and labels
for i, sample in enumerate(chexpert_dataset):
        print(i, sample[0].shape, sample[1].shape)

        ax = plt.subplot(1, 4, i + 1)
        plt.tight_layout()
        ax.set_title(f"Label: {sample[1]}")
        ax.axis('off')
        plt.imshow(sample[0])

        if i == 3:
                plt.show()
                break
```
